```python
# Setup 
%reload_ext autoreload

import sys
sys.path.append("../..")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.z_plotting import plot_scatter
from src.sfms     import fit_ridge_line, lee2015_sfms

df = pd.read_csv("../outputs/tables/master_results.csv")

# Refit SFMS params (fast — just a curve_fit call)
sf_mask     = np.isfinite(df["log_SFR"]) & (df["log_SFR"] > -2.0)
sfms_params, _, _ = fit_ridge_line(
    df.loc[sf_mask, "log_Mstar"].values,
    df.loc[sf_mask, "log_SFR"].values
)
print(f"SFMS params: S0={sfms_params[0]:.3f}, "
      f"a1={sfms_params[1]:.3f}, Mt={sfms_params[2]:.3f}")
```

```python
# SFR vs M*, colour by Gini, 100 kpc pairs
config = {
    "x"         : "log_Mstar",
    "y"         : "log_sSFR",
    "colour_by" : "G",
    "sample"    : "close_pairs_100",
    "sfr_cut"   : -3,
    "show_sfms" : True,
    "title"     : "Gini - Close Pairs 100 kpc",
}
fig, ax = plot_scatter(df, sfms_params, config)
plt.show()
```
