```python
# Diagnostic — Sample composition summary
from src.z_samples import SAMPLE_REGISTRY, sfr_cut
import pandas as pd

rows = []

for name, func in SAMPLE_REGISTRY.items():
    mask        = func(df)
    n_total     = mask.sum()
    n_sf        = (mask & (df["spectral_class"] == "SF")).sum()
    n_with_sfr  = (mask & np.isfinite(df["log_SFR"])).sum()
    n_above_cut = (mask & sfr_cut(df, -2.0)).sum()

    rows.append({
        "sample"         : name,
        "n_total"        : n_total,
        "n_SF_class"     : n_sf,
        "n_finite_SFR"   : n_with_sfr,
        "n_SFR_gt_-2"    : n_above_cut,
    })

summary = pd.DataFrame(rows).set_index("sample")

# Also add galclass breakdown for the full sample
print("=== Sample sizes ===")
print(summary.to_string())

print("\n=== galclass breakdown (full df) ===")
print(df["galclass"].value_counts(dropna=False).sort_index().rename({
    1.0: "1 Ungrouped",
    2.0: "2 Low-mass group",
    3.0: "3 High-mass group",
    4.0: "4 Cluster",
}))

print("\n=== galclass breakdown (SF only) ===")
sf_df = df[df["spectral_class"] == "SF"]
print(sf_df["galclass"].value_counts(dropna=False).sort_index().rename({
    1.0: "1 Ungrouped",
    2.0: "2 Low-mass group",
    3.0: "3 High-mass group",
    4.0: "4 Cluster",
}))

print("\n=== spectral_class breakdown ===")
print(df["spectral_class"].value_counts(dropna=False))

print("\n=== environment breakdown ===")
print(df["environment"].value_counts(dropna=False))
```

```python
# Diagnostic — Wang catalogue diagnostic
import pandas as pd
import numpy as np

BASE = "../../data/catalogs/SAMI_Cindex_data/"

files = {
    "clu_com"  : BASE + "cluout_com_v0.12.txt",
    "clu_sf"   : BASE + "clusterout_sf_hbsn3_v0.12.txt",
    "gama_com" : BASE + "gamaout_com_v0.12.txt",
    "gama_sf"  : BASE + "gamaout_sf_hbsn3_v0.12.txt",
}

dfs = {}
for name, path in files.items():
    # Read with no comment skipping, then strip # from header manually
    raw = pd.read_csv(path, sep=r"\s+", header=0, comment=None)
    raw.columns = [c.lstrip("#") for c in raw.columns]
    dfs[name] = raw
    print(f"=== {name} ===")
    print(f"  rows          : {len(raw)}")
    print(f"  columns       : {list(raw.columns)}")
    print()

# ── ID sets per catalogue ─────────────────────────────────────────────────────
clu_com_ids  = set(dfs["clu_com"]["CATID"].astype(int))
clu_sf_ids   = set(dfs["clu_sf"]["CATID"].astype(int))
gama_com_ids = set(dfs["gama_com"]["CATID"].astype(int))
gama_sf_ids  = set(dfs["gama_sf"]["CATID"].astype(int))

# ── Overlap analysis ──────────────────────────────────────────────────────────
print("=== Cluster catalogue overlaps ===")
print(f"  clu_com only (not in clu_sf)  : {len(clu_com_ids - clu_sf_ids)}")
print(f"  clu_sf only  (not in clu_com) : {len(clu_sf_ids - clu_com_ids)}")
print(f"  in both clu_com AND clu_sf    : {len(clu_com_ids & clu_sf_ids)}")

print()
print("=== GAMA catalogue overlaps ===")
print(f"  gama_com only (not in gama_sf)  : {len(gama_com_ids - gama_sf_ids)}")
print(f"  gama_sf only  (not in gama_com) : {len(gama_sf_ids - gama_com_ids)}")
print(f"  in both gama_com AND gama_sf    : {len(gama_com_ids & gama_sf_ids)}")

print()
print("=== Cross-environment overlaps ===")
all_com_ids = clu_com_ids | gama_com_ids
all_sf_ids  = clu_sf_ids  | gama_sf_ids
print(f"  total unique com galaxies       : {len(all_com_ids)}")
print(f"  total unique sf galaxies        : {len(all_sf_ids)}")
print(f"  com galaxies also in sf         : {len(all_com_ids & all_sf_ids)}")
print(f"  com galaxies NOT in sf          : {len(all_com_ids - all_sf_ids)}")

print()
print("=== Master CSV cross-check ===")
master_ids = set(df["CATID"].astype(int))
print(f"  master_sfms galaxies            : {len(master_ids)}")
print(f"  master in com (not sf)          : {len(master_ids & (all_com_ids - all_sf_ids))}")
print(f"  master in sf                    : {len(master_ids & all_sf_ids)}")
print(f"  master in both com and sf       : {len(master_ids & all_com_ids & all_sf_ids)}")
print(f"  master in neither               : {len(master_ids - all_com_ids - all_sf_ids)}")
```

```python
# Diagnostic — disparity diagnostic between asymmetry_results and master_sfms
import pandas as pd
import numpy as np
import os

# Find project root relative to wherever this notebook lives
# Adjust the number of ".." to match your notebook's depth
PROJECT_ROOT = os.path.abspath("../../")
print("Project root:", PROJECT_ROOT)  # verify this prints the sami/ folder

asym   = pd.read_csv(os.path.join(PROJECT_ROOT, "asymmetry/outputs/tables/asymmetry_results.csv"))
master = pd.read_csv(os.path.join(PROJECT_ROOT, "master/outputs/tables/master_sfms.csv"))

print("=== Row counts ===")
print(f"  asymmetry_results : {len(asym)}")
print(f"  master_sfms       : {len(master)}")

# Align on CATID
asym_ids   = set(asym["catid"].astype(int))
master_ids = set(master["CATID"].astype(int))

print()
print("=== ID overlap ===")
print(f"  in asym only   : {len(asym_ids - master_ids)}")
print(f"  in master only : {len(master_ids - asym_ids)}")
print(f"  in both        : {len(asym_ids & master_ids)}")

# Compare SFR values for shared galaxies
shared_ids = asym_ids & master_ids
asym_shared   = asym[asym["catid"].astype(int).isin(shared_ids)].set_index("catid")
master_shared = master[master["CATID"].astype(int).isin(shared_ids)].set_index("CATID")

# Check what SFR column is called in asym
print()
print("=== asymmetry_results columns ===")
print(list(asym.columns))

print()
print("=== SFR comparison for shared galaxies ===")
# Adjust "log_SFR" in asym if column name differs
if "log_SFR" in asym.columns:
    sfr_diff = (master_shared["log_SFR"] - asym_shared["log_SFR"]).dropna()
    print(f"  mean difference (master - asym) : {sfr_diff.mean():.4f}")
    print(f"  max  difference                 : {sfr_diff.abs().max():.4f}")
    print(f"  galaxies with diff > 0.01       : {(sfr_diff.abs() > 0.01).sum()}")
else:
    print("  log_SFR not found in asym — check column name above")

print()
print("=== SFR range comparison ===")
print(f"  asym   log_SFR : {asym['log_SFR'].min():.2f} to {asym['log_SFR'].max():.2f}")
print(f"  master log_SFR : {master['log_SFR'].min():.2f} to {master['log_SFR'].max():.2f}")

print()
print("=== Mstar range comparison ===")
print(f"  asym   log_Mstar : {asym['log_Mstar'].min():.2f} to {asym['log_Mstar'].max():.2f}")
print(f"  master log_Mstar : {master['log_Mstar'].min():.2f} to {master['log_Mstar'].max():.2f}")

print()
print("=== Suspicious SFR values in master (below -4) ===")
low_sfr = master[master["log_SFR"] < -4]
print(f"  count : {len(low_sfr)}")
if len(low_sfr) > 0:
    print(f"  source_file breakdown:")
    print(low_sfr["source_file"].value_counts())
```
